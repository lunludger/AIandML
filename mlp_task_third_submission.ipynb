### 2. Initialise random weights and biases for the connection between layers

```python
input_size = 5
hidden_size = 4
output_size = 1
np.random.seed(42)
weight_1 = np.random.randn(input_size,hidden_size)
biases_1 = np.random.randn(1,hidden_size)
weight_2 = np.random.randn(hidden_size,output_size)
biases_2 = np.random.randn(1,output_size)
#print(W1,b1,W2,b2)
```

### 3. Use the sigmoid activation function for both the hidden and output layers.

```python
# Sigmoid activation function
def sigmoid(x):
    """Calculates the sigmoid activation function.

    It maps any real-valued input to a value between 0 and 1, making it fit the neural network
    activation.

    Args:
        x (float): The input values

    Returns:
       float: The sigmoid-transformed values
    """
    return 1/(1+np.exp(-x))
```

### 4. Perform forward propagation
Forward propagation of inputs through the network by computing the weighted sums of inputs for each neuron in both the hidden and output layers, then apply the sigmoid activation function to these sums during forward propagation through the network.

```python
def forward(inputs):
    """Performs forward propagation through the multilayer perceptron (MLP).

    The function calculates the weighted sums for the hidden layer and
    output layer, applies the sigmoid activation function to each layer,
    and scales the final output to a quality score between 0 and 10.

    Parameters:
        inputs (numpy.ndarray): A normalized input vector or matrix
                                containing the five basketball player
                                ratings (speed, jump, shooting,
                                intelligence, and strength).

    Returns:
        numpy.ndarray or float:
            - If a single player is provided, returns a quality score
              rounded to one decimal place.
    """

    # Hidden layer
    hidden_input = np.dot(inputs, weight_1) + biases_1
    hidden_output = sigmoid(hidden_input)

    # Output layer
    final_input = np.dot(hidden_output, weight_2) + biases_2
    final_output = sigmoid(final_input)

    # Scale sigmoid output (0-1) to quality score (0-10)
    quality_score = final_output * 10

    return round(float(quality_score.item()), 1)
```

### 5. After implementing the MLP, validate its functionality by passing sample input data through the network and printing the resulting outputs.

```python
# Sample player ratings
X = np.array([90, 85, 92, 88, 80])

# Normalise inputs
player = X / 100

# Predict quality score
quality = forward(player)

print("Basketball Player Ratings")
print("-------------------------")
print("Speed:", 90)
print("Jump:", 85)
print("Shooting:", 92)
print("Intelligence:", 88)
print("Strength:", 80)
print()
print("Predicted Quality:", quality, "/10")
```
