##Sentiment Analysis program using spaCy.
#### The purpose of the notebook is to perform sentiment analysis on Amazon product reviews.

```python
# import necessary libraries
import pandas as pd
import spacy
from spacytextblob.spacytextblob import SpacyTextBlob
```

### 2. Load the Amazon product reviews dataset.

```python
dataframe=pd.read_csv("/content/drive/MyDrive/Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products_May19.csv")
```

### 3.1 Implement a sentiment analysis model using spacy.
### Load the en_core_web_md spaCy

```python
nlp=spacy.load("en_core_web_md")
if "spacytextblob" not in nlp.pipe_names:
    nlp.add_pipe("spacytextblob")
```

### 3.2 Preprocess the text data


3.2.1 Select "reviews.text" column from dataset and retrieve its data

```python
reviews_data = dataframe['reviews.text']
```

3.2.2 Remove all missing values from this column.

```python
clean_data = dataframe.dropna(subset=['reviews.text'])
reviews = clean_data['reviews.text']
```

Removing stop words

```python
def clean_review(review):
  review = str(review).lower().strip()
  doc = nlp(review)
  return " ".join(t.text for t in doc if not t.is_stop and not t.is_punct and not t.is_space)
```

### 3.3 Create a function for sentiment analysis.

```python
def sentiment_analysis(review):
  cleaned = clean_review(review)
  doc = nlp(cleaned)
  polarity = doc._.blob.polarity
  if polarity > 0.1:
    label = "Positive"
  elif polarity <-0.1:
    label = "Negative"
  else:
    label = "Neutral"
  return polarity, label

```

### 3.4 Test model on sample product reviews

```python
print(f"Number of reviews: {len(reviews)}")
for i,r in enumerate(reviews.head(5),1):
  p,s = sentiment_analysis(r)
  print(f"\nReview {i}:")
  print(r[:150]+"...")
  print("Polarity:", round(p,3))
  print("Sentiment:", s)

doc1 = nlp(str(reviews.iloc[0]))
doc2 = nlp(str(reviews.iloc[1]))
print("\nSimilarity:",round(doc1.similarity(doc2),3))

```
