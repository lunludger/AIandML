```python
# semantic.py

# Import the spaCy library
import spacy

# Download the required spaCy models if they are not already installed.
!python -m spacy download en_core_web_md
!python -m spacy download en_core_web_sm



# Load the medium English language model that includes word vectors.
nlp = spacy.load("en_core_web_md")

# Converting the words into spaCy document objects so they can be analysed.
word_1 = nlp("cat")
word_2 = nlp("monkey")
word_3 = nlp("banana")

# Calculating and printing the similarity scores between the words.
print("Similarity between 'cat' and 'monkey':", word_1.similarity(word_2))
print("Similarity between 'cat' and 'banana':", word_1.similarity(word_3))
print("Similarity between 'monkey' and 'banana':", word_2.similarity(word_3))

# Creating spaCy objects for another set of words.
word_4 = nlp("car")
word_5 = nlp("bus")
word_6 = nlp("apple")

# Calculating and printing the similarity scores for the new example.
print("\nMy own example:")
print("Similarity between 'car' and 'bus':", word_4.similarity(word_5))
print("Similarity between 'car' and 'apple':", word_4.similarity(word_6))
print("Similarity between 'bus' and 'apple':", word_5.similarity(word_6))

# Display a heading to separate the results from the two language models.
print("\n--------------------------------------")
print("Using en_core_web_sm")
print("--------------------------------------")

# Load the small English language model.
nlp_sm = spacy.load("en_core_web_sm")

# Convert the same words into spaCy objects using the small model.
word_1_sm = nlp_sm("cat")
word_2_sm = nlp_sm("monkey")
word_3_sm = nlp_sm("banana")

# Calculate and printing the similarity scores using the small model
print("Similarity between 'cat' and 'monkey':", word_1_sm.similarity(word_2_sm))
print("Similarity between 'cat' and 'banana':", word_1_sm.similarity(word_3_sm))
print("Similarity between 'monkey' and 'banana':", word_2_sm.similarity(word_3_sm))
```

###Question 4: Difference between en_core_web_md and en_core_web_sm

Observation

The en_core_web_md model produces more accurate and meaningful similarity scores because it includes pre-trained word vectors that capture semantic relationships between words.

The en_core_web_sm model does not include these word vectors. Instead, it predicts similarity using other linguistic features, making the scores less reliable and sometimes producing warnings.

Therefore
en_core_web_md
uses pre-trained word vectors,
produces more accurate semantic similarity scores.
It is recommended when comparing the meanings of words or documents.
en_core_web_sm does not contain word vectors,
the similarity scores are less accurate and can sometimes be misleading.
It is suitable for basic NLP tasks such as tokenization, part-of-speech tagging, and named entity recognition, but not ideal for semantic similarity tasks.

The similarity scores show that "monkey" and "banana" have a higher semantic similarity than "cat" and "banana". This is because spaCy's word vectors are trained on large text corpora where words that frequently appear in similar contexts develop similar vector representations. Monkeys are commonly associated with bananas in books, articles, and everyday language, so their vectors are closer together in the embedding space.

On the other hand, "cat" and "banana" have a lower similarity score because cats are not commonly associated with bananas. Although both are common nouns, they rarely appear in related contexts, resulting in more distant vector representations.

Similarly, "cat" and "monkey" typically have a moderate similarity score because both are animals and often appear in similar contexts such as discussions about wildlife, pets, or mammals. SpaCy therefore captures both their shared category (animals) and their contextual relationships.

These results demonstrate that spaCy measures semantic similarity based on contextual usage rather than dictionary definitions or explicit category labels. Words that frequently occur in similar contexts or have strong real-world associations receive higher similarity scores, even if they belong to different categories, such as an animal ("monkey") and a fruit ("banana")
